```python
def stateless_gpt(prompt:str)->str:
    from dotenv import load_dotenv
    
    from groq import Groq
    
    load_dotenv()
    
    client = Groq()
    response = client.chat.completions.create(
        model="openai/gpt-oss-120b",
        messages=[
            {
             
                "role":"user",
                "content":prompt
            }
        ]
    
    )
    
    print(response.choices[0].message.content)
```

```python
stateless_gpt("who was pm of uk before 2020?")
```

```python
stateless_gpt("who was before him")
```
